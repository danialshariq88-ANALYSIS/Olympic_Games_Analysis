##### Question : Which Country has the tallest athletes in Latest Olympics? Display Top 5 (Country Names)

```python
noc = pd.read_csv('noc_regions.csv')

data = events.merge(noc[['NOC', 'region']], on='NOC')

latest = data[data['Year'] == data['Year'].max()]

top5 = latest.groupby('region')['Height'].mean().sort_values(ascending=False).head(5)

top5
```

```python
top5.plot(kind='bar', title='Top 5 Countries with Tallest Athletes')
```
